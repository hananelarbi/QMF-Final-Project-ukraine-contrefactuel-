```python
from pathlib import Path
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm
```

```python
print("Current working directory:")
print(os.getcwd())
```

```python
from pathlib import Path

cwd = Path.cwd()

data_path = cwd.parent / "data" / "data_processed"

print("Data path:", data_path)
print("Exists:", data_path.exists())
```

# Data loading 


```python
df = pd.read_csv(
    data_path / "inflation_panel.csv",
    parse_dates=["date"],
    index_col="date"
)

factor = pd.read_csv(
    data_path / "euro_inflation_factor.csv",
    parse_dates=["date"],
    index_col="date"
)

print(df.shape)
print(factor.shape)

df.head()
```

# Merge Data 


```python
ukr = df[["UKR"]].copy()
data = ukr.join(factor, how="inner")

print(data.shape)
data.head()
```

# Regression

```python
Y = data["UKR"]
X = sm.add_constant(data["EURO_FACTOR"])

model = sm.OLS(Y, X).fit()

print(model.summary())
```

```python
data["UKR_counterfactual"] = model.predict(X)

data.head()
```

```python
plt.figure(figsize=(12,6))

plt.plot(data.index, data["UKR"], label="Ukraine actual")
plt.plot(data.index, data["UKR_counterfactual"], label="Counterfactual")

plt.legend()
plt.title("Ukraine Inflation: Actual vs Counterfactual")
plt.show()
```
